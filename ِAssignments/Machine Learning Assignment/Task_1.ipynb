```python
import numpy as np
x = np.array([50, 60, 70, 80, 90]) # Input or feature
y = np.array([150, 180, 210, 240, 270]) # Target variable
```

```python
theta_0 = 0 # W or the weight
theta_1 = 0 #b or the bais
learning_rate = 0.0001
n_iters = 1000
cost_value=[]
for i in range(n_iters):
    y_hat= theta_0*x +theta_1
    error = y_hat-y
    #====================
    D_w = np.mean(error *x)
    D_b = np.mean(error )*1
    #=======================
    theta_0 = theta_0-learning_rate * D_w
    theta_1 = theta_1-learning_rate * D_b
    y_hat = theta_0*x +theta_1
    #cost
    cost = 0.5*np.mean((y_hat - y)**2)
    cost_value.append(cost)
    if (i+1) %100 == 0 or i == 0:
        print(f"n_iters {i +1}: "
            f"cost = {cost}, "
            f"w = {theta_0}"
            f"b ={theta_1}"
                  
                
            )
```

```python
x_new = 70
predicted_price = theta_0 * x_new + theta_1

print(f"The predicted price: {predicted_price:.2f} thousand")
```

```python
import matplotlib.pyplot as plt

# Visualize SSE over iterations
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(range(1, n_iters + 1), cost_value, color='blue', linewidth=2)
plt.title('Cost (SSE) over Iterations')
plt.set_xlabel = plt.xlabel('Iteration')
plt.set_ylabel = plt.ylabel('Cost / SSE')
plt.grid(True)
```
