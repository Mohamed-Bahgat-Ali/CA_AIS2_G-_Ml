```python
#yOUR CODES
import pandas as pd

def display_missing_values(df: pd.DataFrame) -> pd.DataFrame:
    """
    Display a summary table of all columns that contain missing values.

    The table shows:
    1. Column name.
    2. Missing count.
    3. Missing percentage of total rows.
    4. Data type.

    Args:
        df (pd.DataFrame): The input DataFrame.

    Returns:
        pd.DataFrame: A summary table of missing values sorted in descending order.
    """

    # Count missing values per column
    missing_count = df.isnull().sum()

    # Calculate missing percentage
    missing_percent = (missing_count / len(df)) * 100

    # Build summary table
    missing_table = pd.DataFrame({
        'Missing Count': missing_count,
        'Missing Percent (%)': missing_percent.round(2),
        'Data Type': df.dtypes
    })

    # Keep only columns that have missing values
    missing_table = missing_table[missing_table['Missing Count'] > 0]

    # Sort by missing count (highest first)
    missing_table = missing_table.sort_values(by='Missing Count', ascending=False)

    # Display results
    if missing_table.empty:
        print("No missing values found in the dataset.")
    else:
        print(f"Found missing values in {len(missing_table)} columns:")
        print("=" * 60)
        print(missing_table)
        print("=" * 60)
        print(f"Total missing values in dataset: {missing_count.sum():,}")

    return missing_table



missing_info = display_missing_values(df)
```

###  To fix this issue You will fill these values with (Mean or median ) for numeric and (Mode)for categorical

```python
# trying to fill missing values in my datset using for loop 
def fill_missing_values(df: pd.DataFrame) -> pd.DataFrame:
    """
    Fill missing values in a DataFrame using a for loop.

    Rules:
    - Numeric columns (float64, int64) -> fill with Mean.
    - Non-numeric columns (object, str) -> fill with Mode.

    Args:
        df (pd.DataFrame): The input DataFrame with missing values.

    Returns:
        pd.DataFrame: The cleaned DataFrame with no missing values.
    """

    # Make a copy to avoid modifying the original DataFrame
    df_clean = df.copy()

    # Loop over every column in the DataFrame
    for col in df_clean.columns:

        # Check if the column has any missing values
        if df_clean[col].isnull().sum() > 0:

            # Check if the column is NUMERIC (works for int and float)
            if pd.api.types.is_numeric_dtype(df_clean[col]):
                mean_value = df_clean[col].mean()
                df_clean[col] = df_clean[col].fillna(mean_value)
                print(f"[NUMERIC] Filled '{col}' with Mean = {mean_value:.2f}")

            # Otherwise, treat it as CATEGORICAL (works for object and str)
            else:
                mode_value = df_clean[col].mode()[0]
                df_clean[col] = df_clean[col].fillna(mode_value)
                print(f"[CATEGORICAL] Filled '{col}' with Mode = '{mode_value}'")

    print("\n🎉 All missing values have been filled.")
    return df_clean

df_clean = fill_missing_values(df)
df_clean.isnull().sum()

```

```python
### check the category columns that included object
def check_categorical_columns(df: pd.DataFrame) -> None:
    """
    Check and display all categorical (object/string) columns in the DataFrame.

    This function:
    1. Identifies all object/string columns.
    2. Prints the number of unique values in each.
    3. Shows the top 5 most frequent values in each.

    Args:
        df (pd.DataFrame): The input DataFrame.

    Returns:
        None: Prints the results to the screen.
    """

    # 1. Select only object or string columns
    categorical_cols = df.select_dtypes(include=['object', 'str']).columns

    print("=" * 60)
    print(f"Found {len(categorical_cols)} categorical columns.")
    print("=" * 60)

    # 2. Loop through each categorical column
    for col in categorical_cols:
        print(f"\nColumn: {col}")
        print(f"   Unique values: {df[col].nunique()}")
        print(f"   Top 5 values:")
        print(df[col].value_counts().head(5))
        print("-" * 40)


check_categorical_columns(df)
```

```python
##  step  2 ) Incorrect data type as in (( `user_type` &&	`member_gender`)) :  significant change , so  need to clean it.
import pandas as pd

def convert_to_category(df: pd.DataFrame, columns: list) -> pd.DataFrame:
    """
    Convert specified columns to 'category' data type.

    This is useful for columns with a limited number of unique values,
    such as 'user_type' and 'member_gender', to save memory and speed up analysis.

    Args:
        df (pd.DataFrame): The input DataFrame.
        columns (list): A list of column names to convert.

    Returns:
        pd.DataFrame: The DataFrame with converted column types.
    """

    # Make a copy to avoid modifying the original DataFrame
    df_clean = df.copy()

    for col in columns:
        if col in df_clean.columns:
            df_clean[col] = df_clean[col].astype('category')
            print(f"Converted '{col}' to category.")
        else:
            print(f"Column '{col}' not found.")

    
    return df_clean

columns_to_convert = ['user_type', 'member_gender']
df_clean = convert_to_category(df, columns_to_convert)



```

```python
## step 3 ) Good factor for marketing work as in ((`member_birth_year`)) : so need to clean it.
## To know the target audience's age and serve on them

# create a column for users' age depending on the current year
def create_age_column(df: pd.DataFrame, birth_year_col: str = 'member_birth_year') -> pd.DataFrame:
    """
    Create a new 'age' column based on the user's birth year.

    Steps:
    1. Fill any remaining missing values in birth year with the median.
    2. Convert birth year column to integer.
    3. Calculate age based on the year 2019 (the year of the data).

    Args:
        df (pd.DataFrame): The input DataFrame.
        birth_year_col (str): Name of the birth year column.

    Returns:
        pd.DataFrame: The DataFrame with the new 'age' column.
    """

    df_clean = df.copy()

   
    if df_clean[birth_year_col].isnull().sum() > 0:
        median_year = df_clean[birth_year_col].median()
        df_clean[birth_year_col] = df_clean[birth_year_col].fillna(median_year)
        print(f"Filled missing values with median = {median_year:.0f}")
    #=====================================================================
    # Convert to integer
    df_clean[birth_year_col] = df_clean[birth_year_col].astype(int)
    #=====================================================================
    
    df_clean['age'] = 2026 - df_clean[birth_year_col]

    print(f"\nCreated 'age' column.")
    print(f"Age range: {df_clean['age'].min()} to {df_clean['age'].max()}")
    print(f"Average age: {df_clean['age'].mean():.1f}")

    return df_clean
```

```python
## create a function to know the average users ages
def get_average_age(df: pd.DataFrame, age_col: str = 'age') -> float:
    """
    Calculate the average age of users.

    Args:
        df (pd.DataFrame): The input DataFrame.
        age_col (str): Name of the age column. Default is 'age'.

    Returns:
        float: The average age.
    """

    average = df[age_col].mean()

    print(f"Average Age: {average:.2f}")

    return average


# ==========================================
# How to use
# ==========================================
avg_age = get_average_age(df_clean)
```

```python
## another important factor to offer  marketing services is the trip duration time 

## need to change the time from second to minute to be more usable
def convert_duration_to_minutes(df: pd.DataFrame, duration_col: str = 'duration_sec') -> pd.DataFrame:
    """
    Convert trip duration from seconds to minutes.

    Creates a new column 'duration_min' by dividing the seconds column by 60.

    Args:
        df (pd.DataFrame): The input DataFrame.
        duration_col (str): Name of the duration column in seconds. Default is 'duration_sec'.

    Returns:
        pd.DataFrame: The DataFrame with a new 'duration_min' column.
    """

    df_clean = df.copy()

    
    df_clean['duration_min'] = df_clean[duration_col] / 60

    
    df_clean['duration_min'] = df_clean['duration_min'].round(2)

    print(f"Created 'duration_min' column from '{duration_col}'.")
    print(f"Duration range in minutes: {df_clean['duration_min'].min()} to {df_clean['duration_min'].max()}")

    return df_clean

df_clean = convert_duration_to_minutes(df_clean)
print(df_clean[['duration_sec', 'duration_min']].head(10))
```

```python
## I need to change the time from second to Houre  to be more usable

def convert_duration_to_hours(df: pd.DataFrame, duration_col: str = 'duration_sec') -> pd.DataFrame:
    """
    Convert trip duration from seconds to hours.

    Creates a new column 'duration_hr' by dividing the seconds column by 3600.

    Args:
        df (pd.DataFrame): The input DataFrame.
        duration_col (str): Name of the duration column in seconds. Default is 'duration_sec'.

    Returns:
        pd.DataFrame: The DataFrame with a new 'duration_hr' column.
    """

    df_clean = df.copy()

    
    df_clean['duration_hr'] = df_clean[duration_col] / 3600

    
    df_clean['duration_hr'] = df_clean['duration_hr'].round(2)

    print(f"Created 'duration_hr' column from '{duration_col}'.")
    print(f"Duration range in hours: {df_clean['duration_hr'].min()} to {df_clean['duration_hr'].max()}")

    return df_clean
df_clean = convert_duration_to_hours(df_clean)
print(df_clean[['duration_sec', 'duration_hr']].head(10))
```
