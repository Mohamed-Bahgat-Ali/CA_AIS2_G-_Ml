```python
def is_prime(n):
    if n <2:
        return False
    for i in range(2,int(n**0.5)+1):
        if n % i ==0:
            return False
    return True
print('twin prime:')
for i in range(2,1000):
    if is_prime(i)and is_prime(i+2):
        print(f'{i} and {i+2}')

```

```python
def prime_factor1 (n):
    factor=[]
    i  =2 
    while i<=n:
        if n%i == 0:
            factor.append(i)
            n = n //i
        else:
            i+=1
    return factor
    
```

```python
def dec_to_bin(n):
    list = []
    while n > 0:
        i = n%2
        list.append(i)
        n = n//2
    list.reverse()
    return list
     
```

```python
dec_to_bin(11)
```

```python
def perfect_number(x,y):
    
    for i in range (x,y):
        if i <= 0:
            continue
        list_1 = []
        for n in range(1,i):
            if i % n ==0:
                list_1.append(n)
        if sum(list_1) == i:
            print (i)
    
```
